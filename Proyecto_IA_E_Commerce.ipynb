```
ecommerce-ai-project/
│
├── 📁 data/                                  # Gestión de datos (Seguridad RNF-IA-02)
│   ├── 📁 raw/
│   │   └── E-commerce_Customer_Segmentation_2026.csv  # INTACTO (Solo lectura - RF-IA-01/02)
│   ├── 📁 processed/
│   │   ├── clean_customers.csv               # Datos limpios listos para IA (Entregable A3)
│   │   └── data_quality_report.json          # Diagnóstico de nulos/duplicados (Entregable A2)
│   └── 📁 outputs/                           # Contrato de integración (Mitigación R-07)
│       ├── clustering_metrics.json           # Evaluación de las 3 configuraciones + Silhouette (RF-IA-03)
│       ├── segments_summary.json             # Perfiles caracterizados vs. Reglas (RF-IA-04 y RF-IA-05)
│       ├── customer_assignments.csv          # Asignación cliente-segmento descargable (RF-IA-08)
│       └── segments_report.html              # Reporte HTML descargable (RF-IA-08)
│
├── 📁 ai-engine/                             # Motor en Python + Google Colab (Actividades A2, A3, A4, A5)
│   ├── 📁 notebooks/
│   │   ├── 01_Diagnostico_y_Limpieza.ipynb   # Cuaderno Colab (henrrucho + alex: RF-IA-01 y RF-IA-02)
│   │   └── 02_PCA_KMeans_y_Reglas.ipynb      # Cuaderno Colab (Jesús: RF-IA-03, RF-IA-04, RF-IA-05)
│   ├── 📁 src/
│   │   ├── config_columns.py                 # Lista de columnas excluidas para evitar Data Leakage (R-02)
│   │   └── pipeline.py                       # Script automatizado para ejecutar desde consola o API
│   └── requirements.txt                      # Versiones fijas: pandas, scikit-learn, kmodes (R-08)
│
├── 📁 backend-api/                           # Puente de comunicación en Node.js + Express (Paquete 1.3.1)
│   ├── 📁 src/
│   │   ├── 📁 routes/
│   │   │   └── segments.routes.js            # Endpoints de la API
│   │   ├── 📁 controllers/
│   │   │   └── segments.controller.js        # Lógica para leer data/outputs/ y filtrar clientes
│   │   └── server.js                         # Servidor principal de la API REST
│   └── package.json                          # Dependencias fijas de Node.js (express, cors, csv-parser)
│
├── 📁 frontend-dashboard/                    # Interfaz Web en React o Angular (Actividades A6 y A7)
│   ├── 📁 src/
│   │   ├── 📁 components/
│   │   │   ├── FileUploader.jsx              # Carga y validación de estructura CSV (RF-IA-01)
│   │   │   ├── KpiCards.jsx                  # Tarjetas de resumen y métricas Silhouette (RF-IA-04)
│   │   │   ├── SegmentCharts.jsx             # Gráficas de perfiles de compra, gasto y satisfacción (RF-IA-04)
│   │   │   ├── RulesComparison.jsx           # Tabla comparativa IA vs. Reglas RFM (RF-IA-05)
│   │   │   ├── CustomerFilterTable.jsx       # Filtro por segmento y conteo de clientes (RF-IA-06 y RF-IA-07)
│   │   │   └── ExportButtons.jsx             # Botones para descargar CSV y HTML (RF-IA-08)
│   │   ├── 📁 services/
│   │   │   └── api.js                        # Conexión HTTP con tu backend-api
│   │   └── App.jsx                           # Vista principal del Dashboard
│   └── package.json                          # Dependencias del FrontEnd (React/Angular, Chart.js/Recharts)
│
├── 📁 docs/                                  # Entregables documentales (A1, A8, A9)
│   ├── diccionario_datos.md                  # Entregable Semana 3 (OE1)
│   ├── manual_usuario.md                     # Entregable final (Semana 12)
│   └── evidencia_pruebas.md                  # Pruebas en Chrome y Edge (RNF-IA-01 al RNF-IA-05)
│
└── README.md                                 # Instrucciones de instalación y ejecución local
```
